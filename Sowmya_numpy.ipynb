C5. Implement y_pred = X @ w + b for a feature matrix X, weight vector w and bias b. Then compute MSE using y_true. [5 marks]

X @ w + b performs matrix multiplication between the features and weights and then adds the bias. MSE is calculated as the average of the squared differences between y_true and y_pred.

```python
import numpy as np

# Feature matrix
X = np.array([[1, 2], [3,4],[5,6]])

# Weight vector and bias
w = np.array([2, 1])
b = 1

# Prediction
y_pred = X @ w + b

# Actual values
y_true = np.array([6,12,16])

# Mean Squared Error
mse = np.mean((y_true - y_pred) ** 2)

print("Predicted values:", y_pred)
print("MSE:", mse)
```

D1. Data Preprocessing Pipeline [8 marks]

You are given a raw 1D or 2D NumPy array containing NaN values.
Describe, step by step, the NumPy operations you would apply to:
(a) detect missing values,
(b) handle them appropriately and justify your strategy, and
(c) normalize or standardize the clean data before it is fed to a model.

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.


Explanation

Data preprocessing is the process of cleaning and transforming raw data before giving it to a machine learning model.

The pipeline can be divided into three main steps:

(a) Detect missing values

NumPy represents missing numerical values using NaN (Not a Number).

We can detect NaN values using:

np.isnan(data)

It returns True wherever a missing value is present.

To check whether any NaN exists, we can use:

np.isnan(data).any()

(b) Handle missing values

Missing values should be handled before training the model because many ML algorithms cannot work directly with NaN values.

For numerical data, a common strategy is to replace NaN with the mean of the available values.

np.nanmean(data)

nanmean() calculates the mean while ignoring NaN values.

Then we can replace NaN values using:

data[np.isnan(data)] = mean

Why mean imputation?

Mean imputation is simple and suitable when the missing values are relatively few and the numerical data does not contain extreme outliers.

(c) Normalize or standardize the clean data

After handling missing values, the data can be standardized so that features have a comparable scale.

Standardization uses:

 Z =X-mean/Standard deviation



After standardization, the data approximately has mean 0 and standard deviation 1.

For a 2D array, we use axis=0 so that each column/feature is standardized separately

```python
import numpy as np

# Step 1: Create sample data containing NaN
data = np.array([[10, 20], [15, np.nan],[30, 40],[25, 50]], dtype=float)

print("Original data:")
print(data)

# Step 2: Detect missing values
missing = np.isnan(data)

print("\nMissing value locations:")
print(missing)

print("\nAny missing values:", np.isnan(data).any())

# Step 3: Handle missing values using column-wise mean
column_mean = np.nanmean(data, axis=0)

data[np.isnan(data)] = np.take(
    column_mean,
    np.where(np.isnan(data))[1]
)

print("\nData after handling NaN:")
print(data)

# Step 4: Standardize the clean data
mean = np.mean(data, axis=0)
std = np.std(data, axis=0)

standardized = (data - mean) / std

print("\nStandardized data:")
print(standardized)
```

Identify every bug. Explain, in terms of shapes and operator precedence, why each is a problem, and rewrite the corrected code.

There are two main bugs in the given code.

Bug 1: Weight vector has the wrong shape

The input matrix X has shape (3, 2), meaning there are:

3 samples
2 features per sample

Therefore, the weight vector must contain 2 weights, one for each feature.

But the given:

w = np.array([0.5, 1.0, 0.2])

has shape (3,).

The matrix multiplication:

X @ w

requires the number of columns in X to match the size of w.

Here:

X → (3, 2)
w → (3,)

The dimensions 2 and 3 do not match, so matrix multiplication produces a shape error.

It should be:

w = np.array([0.5, 1.0])

Now:

X → (3, 2)
w → (2,)
X @ w → (3,)

Bug 2: Incorrect operator precedence in MSE

The given expression is:

mse = np.mean(y_true - y_pred ** 2)

The problem is that Python evaluates the exponentiation ** before subtraction.

So it is interpreted as:

mse = np.mean(y_true - (y_pred ** 2))

But the Mean Squared Error formula is:

MSE = mean((y_true - y_pred)²)

Therefore, the difference must be calculated first, and then squared:

mse = np.mean((y_true - y_pred) ** 2)


```python
import numpy as np

X = np.array([[1, 2],
              [2, 3],
              [3, 4]])  # shape (3, 2)

w = np.array([0.5, 1.0])  # shape (2,)

b = 2.0

y_pred = X @ w + b

y_true = np.array([10, 20, 30])

mse = np.mean((y_true - y_pred) ** 2)

print("X shape:", X.shape)
print("w shape:", w.shape)
print("y_pred:", y_pred)
print("MSE:", mse)
```

E1. Per-Column Min–Max Normalization [5 marks]

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.


Min-Max normalization scales each feature independently to the range 0 to 1.

The formula is:

X_{norm} = (X-Xmin)/(Xmax-Xmin)

For per-column normalization, we calculate the minimum and maximum separately for each column using axis=0.


```python
import numpy as np

# 2D feature matrix
X = np.array([
    [10, 100, 1000],
    [20, 200, 2000],
    [30, 300, 3000],
    [40, 400, 4000]
])

# Find minimum and maximum of each column
X_min = np.min(X, axis=0)
X_max = np.max(X, axis=0)

# Per-column Min-Max normalization
X_norm = (X - X_min) / (X_max - X_min)

print("Original matrix:")
print(X)

print("\nNormalized matrix:")
print(X_norm)
```
